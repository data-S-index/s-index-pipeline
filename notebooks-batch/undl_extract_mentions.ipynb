```python
import requests
from sindex.sources.undl.client import UNDLClient

url = "https://digitallibrary.un.org/search"
params = {"ln":"en","p":"climate","rm":"wrd","of":"xm","rg":5,"jrec":1}

# plain requests
r1 = requests.get(url, params=params, timeout=60)
print("plain:", r1.status_code, r1.headers.get("Content-Type"), (r1.text[:80] if r1.text else ""))

# your session
c = UNDLClient()
r2 = c.session.get(url, params=params, timeout=60)
print("session:", r2.status_code, r2.headers.get("Content-Type"), (r2.text[:80] if r2.text else ""))
```

```python
from sindex.sources.undl.jobs import extract_undl_mentions

for row in extract_undl_mentions(query="climate", max_records=3):
    print(row["recid"], len(row["dois"]), len(row["urls"]))
    print(row["dois"][:3])
    print(row["urls"][:3])
    print("-----")
```
