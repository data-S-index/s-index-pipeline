## Check DB

```python
con = duckdb.connect(db_path, read_only=True)

con.execute("SHOW TABLES").fetchall()
con.execute("DESCRIBE mdc_index").df()
con.execute("SELECT * FROM mdc_index LIMIT 20").df()
```

```python
con.execute("""
    SELECT *
    FROM mdc_index
    WHERE dataset_norm LIKE 'http%'
    LIMIT 10
""").df()
```

```python
row_count_result = con.execute("SELECT COUNT(*) FROM mdc_index").fetchone()
row_count = row_count_result[0]

print(f"The table has {row_count} rows.")
```
