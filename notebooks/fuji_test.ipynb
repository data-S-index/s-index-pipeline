# F-UJI FAIR score test

Needs a running local f-uji server: 
Installation with conda: 
1) Clone the F-UJI repo: https://github.com/pangaea-data-publisher/fuji
2) Create a python env with the correct Python version e.g.: ```conda create -n fuji python=3.11 -y``` and activate the environment
3) Go to the F-UJI directory in the Anaconda command prompt and install required packages: ```python -m pip install .```
4) Add credentials in ```fuji_server/config/users.py```, e.g., ``` fuji_users = {"bhavesh": "mystrongpassword"} ```
5) Start F-UJI server: ```python -m fuji_server -c fuji_server/config/server.ini```
